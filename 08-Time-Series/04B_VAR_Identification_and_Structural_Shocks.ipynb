**Why this method?**

## 1. The Identification Problem

A VAR residual vector \(u_t\) typically has correlated components, so it cannot be interpreted as independent structural shocks. Identification imposes a mapping \(u_t = B \epsilon_t\), where \(\epsilon_t\) are orthogonal structural shocks and \(B\) encodes economic restrictions.

**Cholesky Ordering:** The most common approach assumes a recursive causal ordering. Variables earlier in the ordering can affect later variables contemporaneously, but not vice versa. This is convenient and transparent, but it is a *strong* economic assumption—changing the ordering changes the shocks and the impulse responses.

# Summary

- VARs are reduced-form systems; their residuals are correlated and **not** structural shocks by default.
- Identification requires economic restrictions to transform residuals into orthogonal shocks.
- Cholesky ordering is a simple, widely used identification scheme—but its implications must be justified.
