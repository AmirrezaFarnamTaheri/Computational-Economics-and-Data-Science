# The Lens: Microfoundations of Aggregates

**What problem are we solving?**
Business cycles—the boom and bust of the economy—were traditionally analyzed using ad-hoc aggregate relationships (like IS-LM). The **Real Business Cycle (RBC)** revolution grounded fluctuations in micro-level optimization. This notebook builds the structural core: equilibrium conditions and the planner's problem that define the model.

**Why this method?**
We build a dynamic stochastic general equilibrium (DSGE) model.
* **Dynamic:** Agents care about the future.
* **Stochastic:** The economy is hit by random shocks.
* **General Equilibrium:** Prices adjust to clear all markets.

This part focuses on the economic structure that later solution methods will linearize and solve.

### 1.1 Equilibrium Conditions

The model consists of the following key equations derived from optimization:

1.  **Household Intratemporal Optimality (Labor Supply):**
    $$ \psi C_t^\sigma L_t^\phi = w_t $$
    The marginal rate of substitution between leisure and consumption equals the real wage.

2.  **Household Intertemporal Optimality (Euler Equation):**
    $$ C_t^{-\sigma} = \beta E_t[ C_{t+1}^{-\sigma} (1+r_{t+1}-\delta) ] $$
    The utility cost of saving one unit today must equal the expected utility gain from consuming the proceeds tomorrow.

3.  **Firm Optimality (Factor Prices):**
    $$ r_t = \alpha A_t K_t^{\alpha-1} L_t^{1-\alpha}, \quad w_t = (1-\alpha) A_t K_t^\alpha L_t^{-\alpha} $$
    Factors are paid their marginal products.

4.  **Resource Constraint:**
    $$ Y_t = C_t + K_{t+1} - (1-\delta)K_t $$
    Output is either consumed or invested.

5.  **Production Function:**
    $$ Y_t = A_t K_t^\alpha L_t^{1-\alpha} $$

6.  **Technology Process:**
    $$ \ln A_t = \rho \ln A_{t-1} + \epsilon_t $$
    Productivity ($A_t$) follows an AR(1) process.

### 1.2 The Social Planner's Problem

By the **First Welfare Theorem**, the competitive equilibrium allocation is identical to the solution of a Social Planner's problem:

$$ \max_{\{C_t, L_t, K_{t+1}\}} E_0 \sum_{t=0}^\infty \beta^t \left( \frac{C_t^{1-\sigma}}{1-\sigma} - \psi \frac{L_t^{1+\phi}}{1+\phi} \right) $$

subject to the resource constraint. This equivalence is computationally useful and theoretically profound: in the absence of frictions (like taxes or sticky prices), the market outcome is Pareto optimal.
