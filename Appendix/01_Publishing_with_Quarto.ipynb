# Introduction

A collection of Jupyter notebooks in a GitHub repository is excellent for development but falls short for presenting a polished course or research project. This "last mile" problem includes significant barriers for end-users, such as environment setup, lack of clear navigation, and a poor reading experience on platforms like GitHub.

**Quarto** is a modern, open-source scientific publishing system designed to solve this problem. It transforms computational documents (like Jupyter notebooks or its native `.qmd` format) into beautiful, accessible, and reproducible websites, books, presentations, and articles. This chapter provides a complete guide to turning a collection of notebooks into a professional website.

### Learning Objectives

By the end of this chapter, you will be able to:

*   **Understand the Quarto Paradigm:** Explain the benefits of a modern scientific publishing system for creating reproducible and professional content.
*   **Master the Quarto Workflow:** Use the Quarto command-line interface (CLI) to initialize, configure, preview, and render a complete website.
*   **Structure a Website:** Skillfully edit the `_quarto.yml` configuration file to control the project's title, navigation, appearance, and content structure.
*   **Leverage Advanced Features:** Use callouts, cross-references, and interactive plots to create sophisticated, polished documents.
*   **Automate Publishing:** Implement a GitHub Actions workflow for continuous deployment, automatically publishing your website upon changes.

## 1. The Quarto Workflow: A Quick Start

The best way to learn Quarto is by building a small project. The workflow centers on a few key commands and one central configuration file.

#### Step 1: Install Quarto
Download and install the Quarto CLI from [quarto.org](https://quarto.org/docs/get-started/).

#### Step 2: Create a Project and Files
In your terminal, create a project directory and the necessary files:
```bash
mkdir my-quarto-site
cd my-quarto-site
touch _quarto.yml index.qmd analysis.ipynb
```

#### Step 3: Add Content
1.  **Configure `_quarto.yml`:** This file is the brain of your site. Add the following content to it. This structure defines the project type, website title, and a navigation sidebar.

    ```yaml
    # _quarto.yml
    project:
      type: website

    website:
      title: "My Data Analysis Project"
      sidebar:
        contents:
          - href: index.qmd
            text: Home
          - href: analysis.ipynb
            text: Data Analysis
    
    format:
      html:
        theme: cosmo
        toc: true
        code-fold: true
    ```

2.  **Create a Homepage (`index.qmd`):** This will be the main landing page.
    ```markdown
    ---
    title: "Welcome to My Project"
    ---

    This website presents the data analysis for my project.
    Please see the **Data Analysis** page for the results.
    ```

3.  **Add a Notebook (`analysis.ipynb`):** Copy the JSON content below into your `analysis.ipynb` file. This is a simple notebook that generates a plot.
    ```json
    {
     "cells": [
      {
       "cell_type": "markdown",
       "metadata": {},
       "source": ["# Data Analysis Results"]
      },
      {
       "cell_type": "code",
       "execution_count": null,
       "metadata": {},
       "outputs": [],
       "source": [
        "import matplotlib.pyplot as plt\n",
        "import numpy as np\n",
        "x = np.linspace(0, 10, 100)\n",
        "plt.plot(x, np.sin(x))\n",
        "plt.title('Sine Wave')\n",
        "plt.show()"
       ]
      }
     ],
     "metadata": {},
     "nbformat": 4,
     "nbformat_minor": 4
    }
    ```

#### Step 4: Preview and Render
Now, use the Quarto CLI to build and view your site:
-   `quarto preview`: Starts a local web server that automatically rebuilds the site when you save a file. This is for development.
-   `quarto render`: Performs a clean, one-time build of the entire website, executing all code. The output is saved to a `_site` directory.

Run `quarto preview` in your terminal. Your browser should open to a fully functional two-page website.

## 3. Automation with GitHub Actions

We can use **GitHub Actions**, a continuous integration and deployment (CI/CD) platform, to automatically render and publish our website. This ensures the live version is always in sync with the repository's `main` branch.

Create a file at `.github/workflows/publish.yml` in your repository with the following content. This workflow checks out your code, installs dependencies, renders the site, and deploys it to GitHub Pages.

```yaml
## .github/workflows/publish.yml
name: Publish Website to GitHub Pages

on:
  push:
    branches: [ main ]

jobs:
  build-and-deploy:
    runs-on: ubuntu-latest
    permissions:
      contents: read
      pages: write
      id-token: write
      
    steps:
      - name: Check out repository
        uses: actions/checkout@v4
        
      - name: Set up Quarto
        uses: quarto-dev/quarto-actions/setup@v2
        
      - name: Install Python and Dependencies
        uses: actions/setup-python@v4
        with:
          python-version: '3.11'
      - name: Install project requirements
        run: pip install -r requirements.txt
          
      - name: Render Quarto Website
        run: quarto render
        env:
          # Ensures matplotlib works in a headless environment
          MPLBACKEND: Agg
        
      - name: Deploy to GitHub Pages
        uses: actions/deploy-pages@v4
```
